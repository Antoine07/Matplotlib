<h2>Matplotlib en un coup d’œil.</h2><p>Matplotlib transforme des valeurs en graphiques, puis permet d’en régler les éléments et d’exporter l’image. pandas prépare les données ; Matplotlib les représente. Seaborn s’appuie sur Matplotlib pour faciliter certains graphiques statistiques. Nous le verrons après les premiers exemples.</p>

<h2>Lire le fichier, vérifier le grain.</h2><p>Une ligne du fichier représente une <strong>ligne de vente</strong>. Une commande peut en contenir plusieurs. Le CSV reprend à l’identique les données du TP Tableau : trois régions et trois catégories, du 1er au 30 janvier 2026.</p><p><code>Ventes</code> est le montant après remise : quantité × prix unitaire × (1 − remise). Exemple : 2 écrans × 220 € × (1 − 0,05) = 418 €. Le fichier ne contient ni coûts ni profits.</p>

```python
import pandas as pd
import matplotlib.pyplot as plt

ventes = pd.read_csv("/workspace/data/commandes.csv", parse_dates=["Date"])

print("Nombre de lignes :", len(ventes))
print("Nombre de commandes :", ventes["Commande"].nunique())
print("Chiffre d'affaires :", ventes["Ventes"].sum())

ventes.head()
```

### Lecture du code et résultat
<code>read_csv</code> ouvre le fichier dans la sandbox Docker ; <code>parse_dates</code> transforme Date en dates. <code>len</code> compte les lignes, <code>nunique</code> les commandes distinctes et <code>sum</code> additionne les ventes. <code>head()</code> montre les cinq premières lignes. Résultat attendu : 60 lignes, 30 commandes, 45 831 €.

<h2>Des lignes aux barres.</h2><p>Pour comparer les ventes par catégorie, il faut d’abord additionner les montants de chaque groupe. Matplotlib reçoit les résultats : <code>barh</code> ne réalise pas cette addition.</p><div class="concept-grid"><article><h3>Figure : <code>fig</code></h3><p>Le document entier, sa taille et son export.</p></article><article><h3>Axes : <code>ax</code></h3><p>Une zone de dessin, avec ses marques, son titre et ses axes gradués.</p></article></div>

```python
par_categorie = ventes.groupby("Categorie")["Ventes"].sum().sort_values()
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.barh(par_categorie.index, par_categorie.values, color="#275c47")
ax.set_title("Ventes par catégorie · 1–30 janvier 2026")
ax.set_xlabel("Ventes après remise (€)")
fig.savefig("/workspace/resultats/01_categories.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>groupby</code> forme les groupes, <code>sum</code> calcule leurs totaux, puis <code>sort_values</code> trie la Series avec ses étiquettes. <code>subplots</code> crée fig et ax ; figsize règle la taille, layout ajuste les marges. <code>barh</code> associe l’index aux montants. La première barre est en bas : le tri croissant place la plus grande en haut. <code>set_title</code> et <code>set_xlabel</code> donnent le contexte. <code>savefig</code> exporte le document dans le dossier prévu par Docker ; <code>show</code> l’affiche. Contrôle : les trois barres totalisent 45 831 €. La taille est en pouces, dpi règle la résolution et bbox_inches évite de couper les textes.

<h2>Un graphique lisible doit être juste.</h2><p>Ajoutons un titre qui répond à la question, une unité, le périmètre et une grille légère. Pour des barres de montants positifs, l’origine à zéro permet de comparer les longueurs.</p>

```python
par_categorie = ventes.groupby("Categorie")["Ventes"].sum().sort_values()
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.barh(par_categorie.index, par_categorie.values, color="#275c47")
ax.set_title("L’informatique domine les ventes · 1–30 janvier 2026")
ax.set_xlabel("Ventes après remise (€) · Toutes régions")
ax.set_xlim(left=0)
ax.grid(axis="x", alpha=0.2)
ax.set_axisbelow(True)
fig.savefig("/workspace/resultats/01_lisible.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>set_xlim(left=0)</code> fixe le début de l’axe horizontal ; <code>grid(axis="x", alpha=0.2)</code> trace une grille légère. <code>set_axisbelow(True)</code> la place derrière les barres. Les données sont identiques à l’exemple précédent ; seule la présentation évolue.

<h2>Quel poids dans l’ensemble ?</h2><p>La part d’une catégorie est son montant divisé par le total du périmètre, multiplié par 100. Ici, Informatique représente 30 776 / 45 831 × 100 ≈ <strong>67,15 %</strong>.</p><p>Le dénominateur fait partie de la définition : « part des ventes de toute la boutique » n’est pas « part des ventes du Sud ».</p>

```python
parts = par_categorie / par_categorie.sum() * 100
print(parts.round(2))
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.barh(parts.index, parts.values, color="#275c47")
ax.set_title("Poids des catégories · 1–30 janvier 2026")
ax.set_xlabel("Part des ventes de toute la boutique (%)")
ax.set_xlim(0, 100)
fig.savefig("/workspace/resultats/01_parts.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
La division applique le même total à chaque montant de la Series. <code>* 100</code> exprime la fraction en pourcentage. <code>round(2)</code> sert à afficher, sans modifier les valeurs utilisées pour le tracé. Les parts non arrondies totalisent 100 % ; un petit écart est possible après arrondi.

<h2>Une autre façon de construire le même graphique.</h2><p>Un histogramme compte les observations dans des intervalles de valeurs. Ici, on étudie les montants des <strong>60 lignes de vente</strong>, pas les paniers des 30 commandes. Le nombre d’intervalles change la finesse de lecture.</p><h3>Exemple Matplotlib</h3>

```python
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.hist(ventes["Ventes"], bins=6, color="#275c47", edgecolor="white")
ax.set_title("Répartition des montants · 60 lignes de vente")
ax.set_xlabel("Montant d’une ligne après remise (€)")
ax.set_ylabel("Nombre de lignes de vente")
fig.savefig("/workspace/resultats/01_histogramme.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Même histogramme avec Seaborn

```python
import seaborn as sns

fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
sns.histplot(data=ventes, x="Ventes", bins=6, ax=ax, color="#275c47")
ax.set_title("Répartition des montants · 60 lignes de vente")
ax.set_xlabel("Montant d’une ligne après remise (€)")
ax.set_ylabel("Nombre de lignes de vente")
fig.savefig("/workspace/resultats/01_seaborn.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>ax.hist</code> reçoit une colonne de nombres ; <code>sns.histplot</code> reçoit la table et le nom du champ. Dans les deux cas, bins=6 demande six intervalles. Avec <code>ax=ax</code>, Seaborn dessine dans notre Axes : les réglages et l’export restent ceux de Matplotlib. Certaines autres fonctions Seaborn agrègent automatiquement : par exemple, barplot estime une moyenne par défaut, pas une somme de ventes.
