Nous distinguons maintenant variation mensuelle, évolution depuis janvier, somme cumulée, moyenne mobile et marge rapportée au chiffre d’affaires.</p><div class="notice"><strong>Un second scénario fictif.</strong><p>Les six mois ci-dessous ne proviennent pas du CSV de janvier : son total et sa période sont différents et il n’a pas de colonne profit. Ce petit tableau indépendant sert à vérifier les calculs à la main, dans le même projet de carnet commercial.</p></div><p>Ouvrez <code>notebooks/02_indicateurs.ipynb</code>. Une ligne représente un mois ; le calendrier est complet et toutes les ventes sont strictement positives.</p>

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "mois": pd.date_range("2026-01-01", periods=6, freq="MS"),
    "ventes": [10000, 12000, 9000, 15000, 18000, 16500],
    "profit": [2000, 3000, 1800, 3750, 4500, 3300]
})
df = df.sort_values("mois").reset_index(drop=True)
mois = df["mois"].dt.strftime("%m/%Y")
df
```

Chaque ligne est un mois ; les dates sont triées. La variable mois contient les étiquettes du graphique. Le profit est un montant fictif fourni, pas calculé depuis le CSV de janvier.

```python
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.plot(mois, df["ventes"], marker="o", color="#275c47")
ax.set_title("Ventes mensuelles · Scénario fictif")
ax.set_ylabel("Ventes (€)")
ax.set_ylim(bottom=0)
fig.savefig("/workspace/resultats/02_mensuel.png", dpi=140, bbox_inches="tight")
plt.show()
```

<h2>Que s’est-il passé depuis le mois précédent ?</h2><p>Variation = (valeur actuelle / valeur précédente − 1) × 100. En février : (12 000 / 10 000 − 1) × 100 = <strong>+20 %</strong>. Le dénominateur change chaque mois.</p>

```python
df["variation_pct"] = df["ventes"].pct_change() * 100
print(df["variation_pct"].round(2))
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.bar(mois, df["variation_pct"], color="#275c47")
ax.axhline(0, color="#203b36", linewidth=1)
ax.set_title("Variation par rapport au mois précédent")
ax.set_ylabel("Variation (%)")
fig.savefig("/workspace/resultats/02_variation.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>pct_change()</code> compare chaque ligne à la précédente et renvoie une fraction ; multiplier par 100 donne un pourcentage. Janvier vaut NaN faute de mois précédent : ce n’est pas 0 %. <code>bar</code> dessine des barres verticales et <code>axhline(0)</code> marque la séparation entre hausse et baisse.

<h2>Garder une référence fixe.</h2><p>Évolution depuis l’origine = (valeur actuelle / valeur initiale − 1) × 100. En mars : (9 000 / 10 000 − 1) × 100 = <strong>−10 %</strong>. La référence est toujours janvier.</p>

```python
df["evolution_pct"] = (df["ventes"] / df["ventes"].iloc[0] - 1) * 100
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.plot(mois, df["evolution_pct"], marker="o", color="#275c47")
ax.axhline(0, color="#203b36", linewidth=1)
ax.set_title("Évolution des ventes depuis janvier")
ax.set_ylabel("Écart à janvier (%)")
fig.savefig("/workspace/resultats/02_evolution.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>iloc[0]</code> prend la première valeur après le tri chronologique. <code>plot</code> relie les mois ; marker rend chaque observation visible. Janvier vaut 0 % car il se compare à lui-même. Cette évolution est parfois dite cumulée : elle ne consiste pas à additionner les variations mensuelles.

<h2>Combien a-t-on vendu depuis le départ ?</h2><p>La somme cumulée additionne les montants depuis janvier jusqu’au mois observé. Fin mars : 10 000 + 12 000 + 9 000 = <strong>31 000 €</strong>. L’unité est l’euro, pas le pourcentage.</p>

```python
df["ventes_cumulees"] = df["ventes"].cumsum()
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.plot(mois, df["ventes_cumulees"], marker="o", color="#275c47")
ax.set_title("Ventes accumulées depuis janvier")
ax.set_ylabel("Ventes cumulées (€)")
ax.set_ylim(bottom=0)
fig.savefig("/workspace/resultats/02_cumul.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>cumsum()</code> conserve une somme progressive dans l’ordre des lignes. Ici, les ventes sont toutes positives : le cumul augmente même quand le montant du mois diminue. Avec des montants nets négatifs, cette propriété ne serait plus garantie.

<h2>Lisser sans effacer les observations.</h2><p>Une moyenne mobile sur trois mois calcule la moyenne du mois courant et des deux précédents. Pour mars : (10 000 + 12 000 + 9 000) / 3 = <strong>10 333,33 €</strong>.</p>

```python
df["moyenne_mobile"] = df["ventes"].rolling(3).mean()
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.plot(mois, df["ventes"], marker="o", label="Ventes", color="#275c47")
ax.plot(mois, df["moyenne_mobile"], marker="s", linestyle="--",
        label="Moyenne sur 3 mois", color="#945324")
ax.set_title("Observer les ventes et leur lissage")
ax.set_ylabel("Ventes (€)")
ax.legend()
fig.savefig("/workspace/resultats/02_moyenne_mobile.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
<code>rolling(3).mean()</code> utilise trois lignes : elles correspondent ici à trois mois consécutifs. Par défaut, il faut trois valeurs ; les deux premiers résultats sont NaN. Deux appels à plot ajoutent deux séries au même Axes. <code>label</code> et <code>legend</code> les identifient ; les marqueurs et tirets complètent la couleur. Un lissage décrit le passé, ce n’est pas une prévision.

<h2>Un ratio de sommes pour l’ensemble.</h2><p>Ici, profit est un montant fictif fourni par mois. Nous nommons précisément l’indicateur : <strong>profit / ventes</strong>, ou marge rapportée au chiffre d’affaires. Ce n’est pas une marge divisée par un coût d’achat.</p><p>En février : 3 000 / 12 000 × 100 = 25 %. Pour le semestre : somme des profits / somme des ventes × 100 = 18 350 / 80 500 × 100 ≈ <strong>22,80 %</strong>.</p>

```python
df["marge_pct"] = df["profit"] / df["ventes"] * 100
marge_globale = df["profit"].sum() / df["ventes"].sum() * 100
print("Marge sur CA du semestre :", round(marge_globale, 2), "%")
fig, ax = plt.subplots(figsize=(8, 4), layout="constrained")
ax.bar(mois, df["marge_pct"], color="#275c47")
ax.set_title("Marge rapportée au chiffre d’affaires · Par mois")
ax.set_ylabel("Profit / ventes (%)")
ax.set_ylim(0, 30)
fig.savefig("/workspace/resultats/02_marge.png", dpi=140, bbox_inches="tight")
plt.show()
```

### Lecture du code et résultat
La division de colonnes calcule le taux de chaque mois. Le ratio des deux sommes calcule le taux global ; chaque mois y pèse selon son chiffre d’affaires. <code>mean()</code> sur les taux donne le même poids à chaque mois : la question traitée est différente.
