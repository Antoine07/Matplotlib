## 1. Charger et vérifier
Le code repère la racine, lit le CSV et prépare le dossier des exports. Résultat attendu : 60, 30, puis 45831.0. Si le chemin manque, conservez data et notebooks dans la même sandbox.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

# Depuis la racine de la sandbox OU le dossier notebooks/corrections.
BASE = Path.cwd()
if not (BASE / "data" / "commandes.csv").is_file():
    BASE = BASE.parent
DATA = BASE / "data" / "commandes.csv"
assert DATA.is_file(), "Ouvrez le notebook depuis la sandbox décompressée."
SORTIES = BASE / "resultats"
SORTIES.mkdir(exist_ok=True)
ventes = pd.read_csv(DATA, parse_dates=["Date"])
print(len(ventes), ventes["Commande"].nunique())
print(ventes["Ventes"].sum())
ventes.head()
```

## 2. Préparer les données
groupby regroupe les lignes ; sum additionne les montants ; sort_values trie les montants avec leur index.

Attendu : Accessoires 1752.50, Mobilier 13302.50, Informatique 30776.00. La somme doit rester 45831.

```python
par_categorie = (
    ventes.groupby("Categorie")["Ventes"]
    .sum()
    .sort_values()
)
print(par_categorie)
```

## 3. Passer à Figure / Axes
fig est le document ; ax est sa zone de tracé. barh reçoit les noms et montants dans le même ordre. Le tri croissant place la plus grande barre en haut.

Le titre donne un constat, les labels donnent les dimensions et l’unité, xlim conserve zéro, grid guide la lecture, suptitle nomme le périmètre. savefig exporte cette Figure à 150 dpi avant show.

**Limite :** il s’agit de ventes, pas de bénéfices.

Prédisez l’image avant d’exécuter.

```python
fig, ax = plt.subplots(figsize=(9, 4), layout="constrained")
ax.barh(par_categorie.index, par_categorie.values, color="#275c47")
ax.set_title("L’informatique domine les ventes de janvier")
ax.set_xlabel("Ventes après remise (€)")
ax.set_ylabel("Catégorie")
ax.set_xlim(left=0)
ax.grid(axis="x", alpha=0.2)
ax.set_axisbelow(True)
fig.suptitle("Boutique fictive · 1–30 janvier 2026 · Toutes régions", fontsize=10)
fig.savefig(SORTIES / "01_categories.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 4. Filtrer avant de regrouper
La comparaison crée un masque ; .loc retient le Sud sans modifier ventes. Reprenez ce principe dans l’exercice 2.

```python
sud = ventes.loc[ventes["Region"] == "Sud"]
par_categorie_sud = sud.groupby("Categorie")["Ventes"].sum().sort_values()
print(par_categorie_sud)
```
