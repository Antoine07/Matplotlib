# Comparer ventes et profit

Un chiffre d’affaires élevé ne suffit pas à décrire le profit. Agréger ventes et profit, puis définir clairement le ratio observé.

**Question métier :** Quelle catégorie associe ventes importantes et meilleur profit rapporté aux ventes ?

Les données sont montées en lecture seule. Enregistrez vos figures dans `/workspace/resultats`.

```python
import pandas as pd
import matplotlib.pyplot as plt

DATA = '/workspace/data/sample_-_superstore.xls'
orders = pd.read_excel(DATA, sheet_name='Orders', engine='xlrd')
orders['Order Date'] = pd.to_datetime(orders['Order Date'])
```

## Notion 1 — Calculer le profit rapporté aux ventes


### Étape 1 — Agréger avant de diviser

Les deux sommes doivent porter sur la même catégorie et le même périmètre.

```python
bilan = orders.groupby('Category')[['Sales','Profit']].sum()
bilan['Taux'] = bilan['Profit'] / bilan['Sales'] * 100
print(bilan.round(2))
```

### Étape 2 — Comparer les taux

Le graphique présente le profit rapporté aux ventes, en pourcentage.

```python
taux = bilan['Taux'].sort_values()
fig, ax = plt.subplots(figsize=(8,4))
ax.barh(taux.index, taux.values)
ax.set(title='Profit rapporté aux ventes', xlabel='Profit / ventes (%)')
plt.show()
```

### Étape 3 — Interpréter

Technology présente le ratio le plus élevé, environ 17,45 %. Ce ratio ne constitue pas un taux de marge commerciale : le coût d’achat n’est pas fourni.

### Exercice 1 — Comparer les taux des sous-catégories

Quelle sous-catégorie présente le plus fort ratio profit/ventes ? Calculez les ratios et représentez les sous-catégories sur un graphique trié.

**À rendre :** Un graphique en pourcentage et le nom de la sous-catégorie présentant le ratio le plus élevé, avec sa valeur.

```python
bilan_sous_categories = orders.groupby("Sub-Category")[["Sales", "Profit"]].sum()
bilan_sous_categories["Taux"] = bilan_sous_categories["Profit"] / bilan_sous_categories["Sales"] * 100
taux = bilan_sous_categories["Taux"].sort_values()
print(taux)
print("Ratio le plus élevé :", taux.idxmax(), taux.max())
fig, ax = plt.subplots(figsize=(9, 6))
ax.barh(taux.index, taux.values, color="#275c47")
ax.axvline(0, color="black", linewidth=1)
ax.set(title="Profit rapporté aux ventes par sous-catégorie", xlabel="Profit / ventes (%)")
plt.tight_layout()
plt.show()
```

### Mise en commun

Labels présente le ratio le plus élevé, environ 43,90 %. Les ventes et les profits sont additionnés par sous-catégorie avant la division. Certains ratios sont négatifs car les profits agrégés correspondants sont négatifs.

## Notion 4 — Comprendre le poids des lignes dans une moyenne


### Étape 1 — Calculer deux indicateurs différents

Le ratio des sommes pondère implicitement les taux par le montant des ventes ; la moyenne ligne à ligne ne le fait pas.

```python
lignes = orders.assign(taux=orders['Profit']/orders['Sales']*100)
ratio = orders.groupby('Region')[['Sales','Profit']].sum()
ratio['Ratio des sommes'] = ratio['Profit']/ratio['Sales']*100
ratio['Moyenne des taux'] = lignes.groupby('Region')['taux'].mean()
print(ratio[['Ratio des sommes','Moyenne des taux']].round(2))
```

### Étape 2 — Visualiser l’écart

Comparer les deux méthodes côte à côte permet d’observer les différences de pondération.

```python
comparaison = ratio[['Ratio des sommes','Moyenne des taux']]
y = range(len(comparaison))
fig, ax = plt.subplots(figsize=(8,4))
ax.barh([i-.18 for i in y], comparaison['Ratio des sommes'], height=.36, label='Ratio des sommes')
ax.barh([i+.18 for i in y], comparaison['Moyenne des taux'], height=.36, label='Moyenne par ligne')
ax.set_yticks(list(y), comparaison.index)
ax.axvline(0, color='black', linewidth=1)
ax.set(title='Deux calculs par région', xlabel='Pourcentage')
ax.legend()
plt.show()
```

### Étape 3 — Interpréter

Dans Central, le ratio des sommes est positif alors que la moyenne des taux ligne à ligne est négative. Les deux agrégations ne répondent pas à la même question.

### Exercice 4 — Comparer les méthodes par catégorie

Dans quelle catégorie le ratio des sommes et la moyenne des taux par ligne s’écartent-ils le plus ? Comparez les deux indicateurs sur un graphique.

**À rendre :** Un graphique comparatif, la catégorie concernée, les deux taux et leur écart en points.

```python
orders["taux_ligne"] = orders["Profit"] / orders["Sales"] * 100
comparaison_categorie = orders.groupby("Category")[["Sales", "Profit"]].sum()
comparaison_categorie["Ratio des sommes"] = comparaison_categorie["Profit"] / comparaison_categorie["Sales"] * 100
comparaison_categorie["Moyenne des taux"] = orders.groupby("Category")["taux_ligne"].mean()
ecarts = (comparaison_categorie["Ratio des sommes"] - comparaison_categorie["Moyenne des taux"]).abs()
print(comparaison_categorie[["Ratio des sommes", "Moyenne des taux"]].round(2))
print("Écart maximal :", ecarts.idxmax(), ecarts.max())
y = range(len(comparaison_categorie))
fig, ax = plt.subplots(figsize=(9, 4))
ax.barh([i + 0.18 for i in y], comparaison_categorie["Ratio des sommes"], height=0.36, label="Ratio des sommes", color="#275c47")
ax.barh([i - 0.18 for i in y], comparaison_categorie["Moyenne des taux"], height=0.36, label="Moyenne des taux", color="#b5cf66")
ax.set_yticks(list(y), comparaison_categorie.index)
ax.set(title="Deux méthodes de calcul par catégorie", xlabel="Taux (%)")
ax.legend()
plt.tight_layout()
plt.show()
```

### Mise en commun

Office Supplies présente le plus grand écart : environ 17,22 % pour le ratio des sommes et 14,00 % pour la moyenne des taux, soit 3,22 points. La moyenne simple donne le même poids à chaque ligne, quel que soit son montant de ventes.
